```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder


dataset = pd.read_csv('1 spring-framework/2015-6.csv')

def plotPerColumnDistribution(df, nGraphShown, nGraphPerRow):
    nunique = df.nunique()
    # pick columns that have between 2 and 49 unique values
    df = df[[col for col in df if 1 < nunique[col] < 50]]
    nRow, nCol = df.shape
    columnNames = list(df)
    nGraphRow = (nCol + nGraphPerRow - 1) // nGraphPerRow   # integer division
    plt.figure(figsize=(6 * nGraphPerRow, 8 * nGraphRow), dpi=80, facecolor='w', edgecolor='k')
    for i in range(min(nCol, nGraphShown)):
        plt.subplot(nGraphRow, nGraphPerRow, i + 1)
        columnDf = df.iloc[:, i]
        if not pd.api.types.is_numeric_dtype(columnDf):
            columnDf.value_counts().plot.bar()
        else:
            columnDf.hist()
        plt.ylabel('counts')
        plt.xticks(rotation=90)
        plt.title(f'{columnNames[i]} (column {i})')
    plt.tight_layout(pad=1.0, w_pad=1.0, h_pad=1.0)
    plt.show()

# plotPerColumnDistribution(dataset, 40, 5)

# Handle missing values: 0 for numbers, empty string for text
num_cols = dataset.select_dtypes(include=np.number).columns
obj_cols = dataset.select_dtypes(exclude=np.number).columns
dataset[num_cols] = dataset[num_cols].fillna(0)
dataset[obj_cols] = dataset[obj_cols].fillna('')

# Encode ONLY text columns, one encoder per column
encoders = {}
for col in ['QualifiedName', 'Name']:
    encoders[col] = LabelEncoder()
    dataset[col] = encoders[col].fit_transform(dataset[col].astype(str))

# Target and features, selected by name
Y = dataset['Complexity'].values
X = dataset.drop(columns=['Complexity']).values
print(dataset.shape)
print(dataset.head())
print(dataset['Complexity'].describe())
```
